### 1c. Variables
A variable is a name for a value, like a label on a sample tube. `=` stores the value under the name, and from then on you use the name instead of the number. A line that starts with `#` is a comment: Python skips it.

```python
weight_kg = 80
height_m = 1.75

bmi = weight_kg / height_m ** 2   # BMI = weight / height squared
print("BMI:", round(bmi, 1))
```

### 1d. A list of values
A list goes in square brackets. Here are four sodium measurements from one patient. `len` counts them, `min` and `max` find the extremes, `sum` adds them up.

### 3b. Loading the dataset
The course dataset is a CSV file of 2,000 **synthetic** admissions to internal-medicine wards. `pd.read_csv` reads it straight from the internet into a table called `df` (short for data frame). `df.shape` gives (rows, columns).

```python
DATA_URL = "https://raw.githubusercontent.com/Rotem-Gershon/cadets-notebooks/main/data/cadets_admissions_release1_core.csv"

df = pd.read_csv(DATA_URL)
df.shape
```

## Part 4 · Asking the dataset questions

### 4a. One column
Pick a column by its name, in square brackets and quotation marks. `.describe()` gives a summary: count, mean, standard deviation, minimum, quartiles and maximum.

```python
df["age"].describe()
```

### 4b. A categorical column
`.value_counts()` counts the admissions in each category.

```python
df["admit_diagnosis_group"].value_counts()
```

### 4c. An outcome rate
In `in_hospital_death`, 1 means the patient died during the admission and 0 means they were discharged alive. The mean of a 0/1 column is the event rate.

```python
death_rate = df["in_hospital_death"].mean()
print("In-hospital death:", round(death_rate * 100, 1), "%")
```

### 4d. Filtering: some of the patients
`df[df["age"] >= 80]` keeps only the rows where the condition is true. The result is a new table, and you can ask it the same questions.

```python
old = df[df["age"] >= 80]

print("Patients aged 80+:", len(old))
print("Death rate:", round(old["in_hospital_death"].mean() * 100, 1), "%")
```

In clinical data not every test is done for every patient. `.isna().sum()` counts the missing values in each column. Why they are missing, and what to do about it, is part of Day 1.

### 4f. A first plot
A histogram of age: how many admissions fall in each age range.

```python
df["age"].hist(bins=30)
plt.xlabel("Age (years)")
plt.ylabel("Number of admissions")
plt.title("Age at admission")
plt.show()
```
