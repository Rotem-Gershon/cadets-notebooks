### 1ג. משתנים
משתנה הוא שם לערך, כמו תווית על מבחנה. `=` שומר את הערך בשם, ואחר כך משתמשים בשם במקום במספר. שורה שמתחילה ב-`#` היא הערה ולא רצה.
